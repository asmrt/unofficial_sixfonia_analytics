```python
#@title ⚙️ ヘッダー・配色の設定
HEADER_TITLE = "回答発表！"  #@param {type:"string"}
HEADER_SUB = "【非公式】雨乃こさめさん コメント単語クイズ(2026.06.23)"  #@param {type:"string"}
THEME = "blue"  #@param ["blue", "pink"]
OUT_PATH = "answer_card.png"  #@param {type:"string"}
```

```python
#@title 🎬 動画リスト（ここを編集）
# video_id: https://youtu.be/XXXXX の XXXXX 部分
# 空リストのセクションは自動でスキップされる

videos = [
    # {"video_id": "qFlAqpIFowQ", "title": "テレパシ / 歌ってみた【雨乃こさめ/シクフォニ】"},
]

shorts = [
    {"video_id": "FAzyNlYQBlI", "title": "ラップバトルキッズです #shorts #人マニア #雨乃こさめ #マイクラ #シクフォニ"},
    {"video_id": "CDJ6kta6FBo", "title": "どしたん？話きこか？【人マニアゲーム】【人マニア】　#shorts"},
]
```

```python
#@title 🖼️ カード画像を生成してダウンロード
from google.colab import files
from sixfonia_analytics.cards import render_video_cards

out = render_video_cards(
    header_title=HEADER_TITLE,
    header_sub=HEADER_SUB,
    sections=[("動画", videos), ("ショート", shorts)],
    theme=THEME,
    out_path=OUT_PATH,
)
files.download(out)
```
